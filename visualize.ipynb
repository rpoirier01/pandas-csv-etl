```python
import pandas as pd
import matplotlib.pyplot as plt
from load_sql import get_recent_high_NO2_locations_as_dataframe
```

```python
df = pd.read_csv("venv/Ozone_Data_Trimmed.csv")
```

```python
df.sort_values("Data Value", ascending=False)
```

## Shows the 10 highest places with Ozone pollution

```python
#Shows the 10 highest places with Ozone pollution
# top_ozone.plot(x="Geo Place Name", y="Data Value", kind="bar")
# # df[df["Data Value"] > 40.00 ].plot(x="Geo Place Name", y="Data Value", kind="bar")
df.sort_values("Data Value", ascending=False).iloc[0:10].plot(x="Geo Place Name", y="Data Value", kind="bar")
plt.xticks(rotation=45)
plt.show()
```

## Data Changes Over Time in a Pivot Table

```python
#Based off of a ChatGPT formatted example:
#Cleanly outputs changes over time
df.groupby(["Geo Place Name", "Time Period"])["Data Value"].max().reset_index().pivot(
    index="Geo Place Name",
    columns="Time Period",
    values="Data Value"
)
```

## Changes in Ozone Levels in West Queens

```python
#Watch the Ozone levels change for specific place over time
df[df["Geo Place Name"] == "West Queens"].sort_values("Time Period").plot(x="Time Period", y="Data Value", kind="line")
plt.xticks(rotation=45)

plt.show()


#.plot(x="Time Period", y="Data Value", kind="line")

# # df[df["Data Value"] > 40.00 ].plot(x="Geo Place Name", y="Data Value", kind="bar")
```
